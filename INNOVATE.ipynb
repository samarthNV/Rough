```python
import openai
import json
from openai import AzureOpenAI

# Use the new client interface
client = AzureOpenAI(
    api_key="83a74d5c9f7a49dd99bf6e1dd64ab961",
    api_version="2024-02-01",
    azure_endpoint="https://innovate-openai-api-mgt.azure-api.net/innovate-tracked/deployments/gpt-4o-mini/chat/completions?api-version=2024-02-01",
)

tickets_string = json.dumps(tickets, indent=4)
# Initial system message to set assistant behavior
messages = [
    {"role": "system", "content": "You are a helpful assistant and guide the user to find out the ticket they need to raise in order to resolve the error they are facing - " + tickets_string}
]

print("Start chatting with the assistant (type 'exit' to stop):\n")

while True:
    question = input("You: ")

    if question.lower() in ["exit", "quit"]:
        print("Ending the chat. Goodbye!")
        break

    # Append user message to conversation history
    messages.append({"role": "user", "content": question})

    # Send request
    response = client.chat.completions.create(
        model="gpt-4o-mini",  # Use your deployment name
        messages=messages
    )

    # Extract assistant reply
    reply = response.choices[0].message.content
    print("Assistant:", reply, "\n")

    # Append assistant reply to messages
    messages.append({"role": "assistant", "content": reply})

```
